## What is Time Series?

Time Series is data collected over time in a sequence.

- The order of data matters
- Each value depends on time

### Examples:
- Daily temperature
- Stock prices
- Sensor readings

## What is Time Series Data?

Time Series Data is a dataset where observations are recorded at regular time intervals.

### Examples of intervals:
- Hourly
- Daily
- Monthly
- Yearly

**Key Point:**  
Time is the most important feature in this type of data.

## Cross-Sectional Data

Data collected at a single point in time across multiple entities.

- No time dependency

### Examples:
- Income of 100 people in 2024
- Marks of students in one exam

## Pooled (Panel) Data

Combination of Time Series and Cross-Sectional Data.

- Same entities tracked over time

### Examples:
- Salary of employees over multiple years
- Sales of multiple stores over months

## Types of Data Comparison

| Data Type            | Description                              |
|---------------------|------------------------------------------|
| Cross-Sectional     | Multiple entities at one time point      |
| Time Series         | One entity over multiple time points     |

## Time Series Models

Time Series Models are used to learn patterns from past data and predict future values.



## Types of Time Series Models

### 1. Statistical Models
- AR (Autoregressive – uses past values)
- MA (Moving Average – uses past errors)
- ARMA (combines AR + MA for stationary data)
- ARIMA (handles non-stationary data using differencing)
- ARIMAX (ARIMA with external variables)
- SARIMA (handles seasonality)
- SARIMAX (SARIMA with external/exogenous variables)



### 2. Machine Learning Models
- XGBoost (tree-based boosting model using lag features)


### 3. Deep Learning Models
- RNN (Recurrent Neural Network – basic sequence model)
- LSTM (captures long-term dependencies)
- GRU (efficient version of LSTM)
- Autoencoder (learns compressed representations for anomaly detection)
- VAE (Variational Autoencoder – probabilistic representation learning)



### 4. Modern Tools / Frameworks
- Prophet (handles trend and seasonality automatically)

## What is Seasonality?

Seasonality means a pattern that repeats at regular time intervals.

This pattern occurs because of external factors like:
- Weather
- Festivals
- Business cycles

### Example:
- Ice cream sales increase every summer
- E-commerce sales increase during festivals (Diwali, Christmas)

### Key Idea:
If something repeats after a fixed time period → it is seasonality



### How to Identify Seasonality?
- Repeated peaks and drops at fixed intervals
- Same pattern every month / year



### Real-Life Example:
| Metric | Jan | Feb | Mar | Apr | May | Jun |
|--------|-----|-----|-----|-----|-----|-----|
| Month  | Jan | Feb | Mar | Apr | May | Jun |
| Sales  | 100 | 120 | 150 | 200 | 250 | 300 |




| Metric | Jul | Aug | Sep | Oct | Nov | Dec |
|--------|-----|-----|-----|-----|-----|-----|
| Month  | Jul | Aug | Sep | Oct | Nov | Dec |
| Sales  | 320 | 300 | 250 | 200 | 180 | 150 |

Pattern repeats every year → Seasonality

## What is Stationary Data?

A time series is called stationary if its statistical properties do not change over time.

These properties include:
- Mean (average)
- Variance (spread)
- Autocorrelation (relationship between values)



### Simple Understanding:
Stationary data = stable behavior over time



### Non-Stationary Data:
- Trend present (increasing or decreasing)
- Changing variance
- Seasonality present



### Example:

Non-Stationary:
Sales: 100, 120, 150, 200, 260 (increasing trend)

Stationary:
Values fluctuate around same level:
100, 105, 98, 102, 101



### Why Stationarity is Important?
Most models like ARIMA require stationary data to work properly.


<img src="Stationary_nonStationary.png" alt="Time Series info" width="1000" height="600" />


## How to Check Stationarity?

### 1. Visual Method
Plot the data:
- If trend or seasonality is visible → non-stationary
- If stable → stationary



### 2. Statistical Test (ADF Test)
Augmented Dickey-Fuller Test

- Null Hypothesis: Data is non-stationary
- If p-value < 0.05 → Data is stationary



### 3. Rolling Statistics
Check:
- Rolling Mean
- Rolling Variance

If they are constant → stationary

## ACF (Autocorrelation Function)

ACF measures correlation between:
current value and past values

Example:
How today's value is related to yesterday, day before yesterday, etc.



## PACF (Partial Autocorrelation Function)

PACF measures:
direct relationship between current value and lag,
removing influence of intermediate lags

* Lag = How many steps back in time you're looking.


### Simple Difference:
ACF → total correlation  
PACF → direct correlation  



### Why Important?
Used to identify:
- AR order (from PACF)
- MA order (from ACF)

## AR (Autoregressive Model)

AR model predicts future values using past values.

### Formula Idea:
Current value depends on previous values

Example:
Y(t) = c + a1*Y(t-1) + a2*Y(t-2)

- Y(t) → "Today's Value" (The Goal)
- c → "Base Value" (The Starting Point)
- a1, a2 → "Weights" (The Importance)
- Y(t−1) → "Yesterday's Value" (Lag 1)
- Y(t-2) → "Day Before Yesterday's Value" (Lag 2)

### Simple Example:
Today's temperature depends on yesterday's temperature



### Key Point:
Uses past values only

## MA (Moving Average Model)

MA model uses past errors to make predictions.

### Idea:
Current value depends on past prediction errors



### Example:
If yesterday’s prediction had error,
today’s value adjusts using that error



### Key Point:
Uses past errors, not past values

<img src="MA_working.png" alt="Time Series info" width="700" height="500" />


## ARMA Model

Combination of:
- AR (past values)
- MA (past errors)



### Use Case:
Works only for stationary data



### Idea:
Uses both past values and past errors

<img src="ARMAworking_G.png" alt="Time Series info" width="900" height="500" />


## ARIMA Model

ARIMA = AR + I + MA

Where:
- AR = past values
- I = differencing (to make data stationary)
- MA = past errors



### Why ARIMA?
Real-world data is usually non-stationary

So we:
* Step 1: Make data stationary
* Step 2: Apply AR + MA



### Example:
Sales data with trend → use ARIMA
